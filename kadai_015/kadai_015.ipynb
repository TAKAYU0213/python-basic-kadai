```python
# 人間を管理するクラス
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を返すメソッド
    def get_name_and_age(self):
        return self.name, self.age


# インスタンス化
human = Human("田中", 25)

# 属性を出力
print("名前:", human.name)
print("年齢:", human.age)
```
