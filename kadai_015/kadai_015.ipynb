```python
# 人間を管理するクラス
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を返すメソッド
    def printinfo(self):
        print(f"名前: {self.name}, 年齢: {self.age}")
        return

person1 = Human("太郎", 25)
person1.printinfo()        
```
