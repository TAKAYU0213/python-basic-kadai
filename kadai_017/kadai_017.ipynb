```python
# 人間を管理するクラス
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を返すメソッド
    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人です。")
        else:
            print(f"{self.name}は大人ではありません。")
        return

person1 = Human("太郎", 25)
person1.check_adult()        
```
